```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
import warnings

warnings.filterwarnings('ignore')
```

```python
data=pd.read_csv("data/WA_Fn-UseC_-Telco-Customer-Churn (1).csv")
```

```python
corr=data.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    annot_kws={'size':6},
    cmap='coolwarm'
)
```

```python
data=data.replace({
    "Yes":1,
    "No":0
})
data
```

```python
def preprocessing(data):
    data['gender']=data['gender'].replace({'Male':1,'Female':0})
    ohe=['Contract','InternetService','PaymentMethod']
    data=pd.get_dummies(data,columns=ohe,drop_first=True)

    return data
```

```python
data=preprocessing(data)
```

```python
data.dtypes
```

```python
data=data.replace({'No internet service':0})
data=data.replace({'No phone service':0})
```

```python
data
```

```python
data.dtypes
```

```python
data['TotalCharges']
```

```python

data['TotalCharges'] = pd.to_numeric(data['TotalCharges'], errors='coerce')
data['TotalCharges'] = data['TotalCharges'].fillna(0)
```

```python
x=data.drop(['customerID','Churn'],axis=1)
y=data['Churn']
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.3,random_state=42)
```

```python
from sklearn.metrics import classification_report
def model_training(model,x_train,y_train,x_test,y_test):
    model.fit(x_train,y_train)
    print(model.score(x_train,y_train))

    y_pred=model.predict(x_test)
    print(classification_report(y_test,y_pred))
```

```python
from sklearn.model_selection import GridSearchCV

params={'criterion':['gini','entropy'], 
        'splitter':['best','random'],
        'max_depth':[None,4,6,7,9],
        'min_samples_split':[2,4,6,8], 
        'min_samples_leaf':[1,2,3,5],
        'random_state':[42,90]
       }
grid=GridSearchCV(
    estimator=DecisionTreeClassifier(),
    param_grid=params,
    cv=5,
    n_jobs=-1
)
grid.fit(x_train,y_train)
grid.best_params_
```

```python
dt=DecisionTreeClassifier(
    criterion= 'entropy',
    max_depth=6,
    min_samples_leaf= 3,
    min_samples_split= 2,
    random_state= 42,
    splitter='best',
    class_weight='balanced'
)
model_training(dt,x_train,y_train,x_test,y_test)
    
```

```python
params={
        'n_estimators':[101,151],
        'max_depth':[None,4,6,7,9],
        'min_samples_split':[2,5], 
        'min_samples_leaf':[1,2,3],
       }
grid=GridSearchCV(
    estimator=RandomForestClassifier(),
    param_grid=params,
    cv=3,
    n_jobs=-1
)
grid.fit(x_train,y_train)
grid.best_params_
```

```python
Logr=LogisticRegression(max_iter=10000,class_weight='balanced')
model_training(Logr,x_train,y_train,x_test,y_test)
```

```python
rm = RandomForestClassifier(
    max_depth= 7,
    min_samples_leaf= 3,
    min_samples_split= 2,
    n_estimators= 151,
    random_state=42
)
model_training(rm,x_train,y_train,x_test,y_test)
```
